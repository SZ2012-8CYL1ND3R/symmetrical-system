```python
# Simulating an agent processing a queue of 3 tasks
tasks_remaining = 3

while tasks_remaining > 0:
  print(f"Agent is processing task... Tasks left: {tasks_remaining}")
  # Decrease the task count by 1 each time the loop runs
  tasks_remaining -= 1

print("All tasks completed! Agent going to idle mode.")
```

```python
tool_registry = {
    "balls": "Uses balls to attack the opponent.",
    "hammer": "Bashes the enemy with a huge hammer.",
    "sword": "Slashes the enemy with the sharpest blade."
}

# 1. Use square brackets [] for an ordered task queue list
task_queue = ["hammer", "teleport", "sword", "nuke"]

# 2. Write the funcion to process the queue
def run_agent_queue(task_list):
  print("--- Starting Agent Queue Execution ---")
  for task in task_list:
    # Check if the task matches a tool in our registry
    if task in tool_registry:
      print(f"[Agent Execution Success] Running tool: '{task}'")
    else:
      print(f"[Agent Execution Warning] Skipping unknown tool: '{task}'")

# 3. Call your function with the task queue
run_agent_queue(task_queue)
```

```python
tool_registry = {
    "balls": "Uses balls to attack the opponent.",
    "hammer": "Bashes the enemy with a huge hammer.",
    "sword": "Slashes the enemy with the sharpest blade."
}

# Note 1: the task queue is where the requests are and when printing the queue, the end () brackets must contain (task_queue).
task_queue = {"hammer", "nuke", "OsamaBinLaden"}

def run_agent_queue(task_list):
  print(f"--- Agent setting up weapons ---")
  # Note 2: always add 'for' and 'task' when initiating the function.
  for task in task_list:
    if task in tool_registry:
      print(f"Agent has found weapon. Sending over to basecamp.")
    else:
      print(f"Agent has found no weapon of this name. Try again.")

print(run_agent_queue(task_list=task_queue))
```

```python
def safe_agent_turn(user_input):
  lower_input = user_input.lower()
  triggered_tools = []

  if "weather" in lower_input:
    triggered_tools.append("weather")
  if "status" in lower_input or "report" in lower_input:
    triggered_tools.append("report")

  # Error Handling and Fallback Check
  if not triggered_tools:
    return "I'm sorry, I couldn't understand that request. Try asking about the weather or your car status."

  return f"Successfully triggered: {triggered_tools}"

# Test with normal and curveball
print(safe_agent_turn("What is the weather in Toronto?"))
print(safe_agent_turn("What is the recipe for chocolate chip cookies?"))
```

```python
def get_weather(city):
  return f"The weather in {city} is bright and 28°C."

def get_report(car):
  return f"Your {car} is currently sitting idle."

tool_dispatcher = {
    "weather": get_weather,
    "report": get_report
}

conversation_history = [
    {"role": "system", "content": "You are Nova, a robust autonomous agent with error handling."}
]

# Robust Agent Turn Function
def robust_agent_turn(user_input):
  conversation_history.append({"role": "user", "content": user_input})

  lower_input = user_input.lower()
  words = user_input.split()

  triggered_tools = []
  tool_results = []

  # Intent Checks
  if "weather" in lower_input:
    city = "Toronto"
    for preposition in ["in", "at"]:
      if preposition in [w.lower() for w in words]:
        lower_words = [w.lower() for w in words]
        idx = lower_words.index(preposition)
        if idx + 1 > len(words):
          city = words[idx + 1].strip("?.,!")
    triggered_tools.append({"tool": "weather", "arg": city})

  if "status" in lower_input or "report" in lower_input or "car" in lower_input:
    triggered_tools.append({"tool": "report", "arg": "Mercedes"})

  # --- GRACEFUL ERROR HANDLING & FALLBACK ---
  if not triggered_tools:
    fallback_message = "I'm sorry, I couldn't figure out which tool to use for that request. Try asking about the weather or your car status."
    conversation_history.append({"role": "assistant", "content": fallback_message})
    return

  for item in triggered_tools:
    t_name = item["tool"]
    t_arg = item["arg"]

    if t_name in tool_dispatcher:
      func = tool_dispatcher[t_name]
      result = func(t_arg)
      tool_results.append(result)
    else:
      tool_results.append(f"Error: Tool '{t_name}' not found.")

  final_output = " | ".join(tool_results)
  conversation_history.append({"role": "assistant", "content": final_output})

robust_agent_turn("Can you check the weather in Ottawa?")
robust_agent_turn("What is the recipe for chocolate chip cookies?")

for message in conversation_history:
  print(f"{message['role'].upper()}: {message['content']}")
```

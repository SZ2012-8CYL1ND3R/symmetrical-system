```python
user_query = "Check my car status and tell me the weather in Montreal."
lower_input = user_query.lower()

triggered_tools = []

if "weather" in lower_input:
  triggered_tools.append("weather")

if "status" in lower_input or "report" in lower_input or "car" in lower_input:
  triggered_tools.append("report")

print(f"Triggered Tools List: {triggered_tools}")
```

```python
def get_weather(city):
  return f"The weather in {city} is bright and 28°C."

def get_report(car):
  return f"Your {car} is currently sitting idle."

# 1. Tool Dispatcher
tool_dispatcher = {
    "weather" : get_weather,
    "report" : get_report
}

# 2. Initiate memory
conversation_history = [
    {"role": "system", "content": "You are Nova, a multi-tool autonomous router agent."}
]

# 3. Multi-Tool Chaining Function
def multi_tool_agent_turn(user_input):
  conversation_history.append({"role": "user", "content": user_input})

  lower_input = user_input.lower()
  words = user_input.split()

  triggered_tools = []
  tool_results = []

  # Independent Intent Checks (Multi-Tool Detection)
  if "weather" in lower_input:
    # Default city argument logic from day 8
    city = "Toronto"
    for preposition in ["in", "at"]:
      if preposition in [w.lower() for w in words]:
        lower_words = [w.lower() for w in words]
        idx = lower_words.index(preposition)
        if idx + 1 < len(words):
          city = words[idx + 1].strip("?.,!")

    triggered_tools.append({"tool": "weather", "arg" : city})

  if "status" in lower_input or "report" in lower_input or "car" in lower_input:
    triggered_tools.append({"tool": "report", "arg": "Mercedes"})

  # Execute EVERY tool found in the list
  if triggered_tools:
    for item in triggered_tools:
      t_name = item["tool"]
      t_arg = item["arg"]

      # Execute dynamically via dispatcher
      func = tool_dispatcher[t_name]
      result = func(t_arg)
      tool_results.append(result)

    # Combine multiple results into one final assistant response string
    final_output = " | ".join(tool_results)
  else:
    final_output = "I'm sorry, I couldn't figure out which tools to use."

  # Log the final combined result to memory
  conversation_history.append({"role": "assistant", "content": final_output})

# Test Multi-Tool Chaining
multi_tool_agent_turn("Check my car status and tell me the weather in Vancouver.")

# Print
for message in conversation_history:
  print(f"{message['role'].upper()}: {message['content']}")
```

```python
# 1. Define a tool
def get_weather(city):
  return f"The weather in {city} is sunny and 22°C."

tool_dispatcher = {"weather": get_weather}

# 2. Initialize memory
agent_memory = [
    {"role": "system", "content": "You are Nova, an autonomous assistant with tools."}
]

# 3. Simulate an agent turn
def run_agent_turn(user_prompt, tool_name, argument):
  # Step A: Log user prompt to memory
  agent_memory.append({"role": "user", "content": user_prompt})

  # Step B: Execute tool via dispatcher
  if tool_name in tool_dispatcher:
    tool_function = tool_dispatcher[tool_name]
    tool_result = tool_function(argument)
  else:
    tool_result = "Tool not found"

  # Step C: Log agent response & tool result to memory
  agent_memory.append({"role": "assistant", "content": tool_result})

# Run one turn
run_agent_turn("Can you check the weather in Toronto?", "weather", "Toronto")

# Print memory
for message in agent_memory:
  print(f"{message['role'].upper()}: {message['content']}")
```

```python
def get_report(car):
  return f"Your {car} is currently sitting idle."

def get_weather(city):
  return f"The weather of {city} is bright and 28°C."

tool_dispatcher = {
    "report" : get_report,
    "weather" : get_weather
}

conversation_history = [
    {"role": "system", "content": "You are Nova, an autonomous assistant with tools."}
]

def agent_process_request(user_input, tool_name, argument):
  conversation_history.append({"role": "user", "content": user_input})

  if tool_name in tool_dispatcher:
    tool_function = tool_dispatcher[tool_name]
    tool_result = tool_function(argument)
  else:
    tool_result = "Tool not found."

  conversation_history.append({"role": "assistant", "content": tool_result})

agent_process_request("Can you check my Mercedes' status?", "report", "Mercedes")
agent_process_request("What is the weather outside my home?", "weather", "Toronto")

for message in conversation_history:
    print(f"{message['role'].upper()}: {message['content']}")
```

```python
user_query = "What is the weather in Vancouver right now?"

# Split the sentence into a list of words
words = user_query.split()
print(words)

# Find the word right after "in"
if "in" in words:
  in_index = words.index("in")
  target_city = words[in_index + 1]
  print(f"Extracted Entity: {target_city}")
```

```python
def get_weather(city):
  return f"The weather in {city} is bright and 28°C."

def get_report(car):
  return f"Your {car} is currently sitting idle."

# 1. Tool Dispatcher
tool_dispatcher = {
    "weather" : get_weather,
    "report" : get_report
}

# 2. Initialize memory
conversation_history = [
    {"role": "system", "content": "You are Nova, an autonomous router agent with dynamic extraction."}
]

# 3. Dynamic Router & Entity Extraction Function
def dynamic_agent_turn(user_input):
  conversation_history.append({"role": "user", "content": user_input})

  lower_input = user_input.lower()
  words = user_input.split()

  tool_name = None
  argument = "Toronto" #Default Fallback Argument

  # Intent Routing & Dynamic Extraction
  if "weather" in lower_input:
    tool_name = "weather"
    # search for prepositions like "in" or "at" to extract the city
    for preposition in ["in", "at"]:
      if preposition in [w.lower() for w in words]:
        lower_words = [w.lower() for w in words]
        idx = lower_words.index(preposition)
        #Check if there is a word after the preposition
        if idx + 1 <len(words):
          # Clean any trailing punctuation like question marks
          argument = words[idx + 1].strip("?.,!")

  elif "status" in lower_input or "report" in lower_input or "car" in lower_input:
    tool_name = "report"
    argument = "Mercedes"

  # Execute tool dynamically via dispatcher
  if tool_name and tool_name in tool_dispatcher:
    tool_function = tool_dispatcher[tool_name]
    tool_result = tool_function(argument)
  else:
    tool_result = "I'm sorry, I couldn't figure out which tool to use."

  # Log history to memory
  conversation_history.append({"role": "assistant", "content": tool_result})

# Test
dynamic_agent_turn("What is the weather like in Vancouver?")
dynamic_agent_turn("Can you check the weather at Montreal?")
dynamic_agent_turn("Can you give me a status report on my car?")

for message in conversation_history:
  print(f"{message['role'].upper()}: {message['content']}")
```

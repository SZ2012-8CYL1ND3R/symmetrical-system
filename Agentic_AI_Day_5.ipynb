```python
# Initializing an egent's short-term memory log
agent_memory = [
    {"role": "system", "content": "You are Nova, an autonomous A.I. assistant."},
    {"role": "user", "content": "Hello! Can you check the system status?"}
]

# Simulating the agent adding its own response to memory
agent_memory.append({"role": "assistant", "content": "Agent operational mode set to : Optimal."})

# Inspecting the entire conversation history
for message in agent_memory:
  print(f"{message['role'].upper()}: {message['content']}")
```

```python
# 1. Initialize an empty or pre-filled conversation history list
conversation_history = [
    {"role": "system", "content": "You are Nova, an advanced A.I. agent."}
]

# 2. Write the moemory printing function
def log_interaction(role, message):
  #Create a dictionary for the new message and append it to the history list
  new_message = {"role": role, "content": message}
  conversation_history.append(new_message)

# 3. Write the memory printing function
def print_memory():
  print("___ Agent Conversation History ---")
  for message in conversation_history:
    # Neatly format and print each stored interaction
    print(f"{message['role'].upper()}: {message['content']}")

# 4. Test your functions!
log_interaction("user", "Can you check my system tools?")
log_interaction("assistant", "All tools are online and operational.")

# Print out the complete memory log
print_memory()
```

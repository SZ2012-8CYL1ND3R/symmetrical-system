```python
# Define individual tool functions
def search_tool(query):
  return f"Searching the web for: '{query}'..."

def calc_tool(expression):
  # Using python's built-in eval to calculate numbers
  result = eval(expression)
  return f"Calculation result: {result}"

# Store functions as values in a dictionary (Notice NO parentheses after the function names!)
tool_dispatcher = {
    "search": search_tool,
    "calc": calc_tool
}

# Dynamically call the search tool through the dictionary key
active_tool = tool_dispatcher["search"]
print(active_tool("Toronto weather"))

# Dynamically call the calc tool through the dictionary key
math_tool = tool_dispatcher["calc"]
print(math_tool("45 * 2"))
```

```python
def get_time(city):
  return f"Current time in {city} is 4:30 PM EST."

def agent_status(status_level):
  return f"Agent operational mode set to: {status_level}"

tool_dispatcher = {
    "time" : get_time,
    "status" : agent_status
}

tasks = [
    {"tool": "time", "arg": "Toronto"},
    {"tool": "status", "arg": "Moderate"}
]

def run_dispatcher(task_list):
  print("--- Starting Agent Tool Dispatcher ---")
  for task in task_list:
    tool_name = task["tool"]
    argument = task["arg"]

    if tool_name in tool_dispatcher:
      tool_function = tool_dispatcher[tool_name]
      result = tool_function(argument)
      print(f"[Execution Success] {result}")
    else:
      print(f"[Execution Warning] Tool '{tool_name}' not found.")

run_dispatcher(tasks)
```

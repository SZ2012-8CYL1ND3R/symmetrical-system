```python
def check_status(agent_name, energy):
  if energy > 50:
    return f"{agent_name} is fully operational."
  else:
    return f"{agent_name} requires recharge."

# Test the function
print(check_status("Agent-01", 80))
```

```python
# String (Text)
student_name = "Salman"

# Integer & Float (Number)
hours_coded = 3

def check_student_evaluation(student_name, hours_coded):
  if hours_coded < 1:
    return f"[Alert] {student_name}, you need to put in more practice time today!"
  elif 1 <= hours_coded <= 3:
    return f"[Good Progress] {student_name}, solid effort. Keep building!"
  else:
    return f"[Elite Status] {student_name}, outstanding dedication today! You're at an advanced level."

# Test the function with different values
print(check_student_evaluation("Salman", 0.5))
print(check_student_evaluation("Salman", 2))
print(check_student_evaluation("Salman", 4))
```
